```python
%%javascript

class Pet {
  constructor(name) {
    this.name = name;
    this.hunger = 67;
    this.energy = 90;
    this.happiness = 100;
  }

  feed() {
    this.energy += 50;
    this.hunger -= 25;
    this.happiness += 20;
    console.log(`${this.name} has been fed!`);
  }

  play() {
    this.energy -= 30;
    this.hunger += 60;
    this.happiness += 75;
    console.log(`${this.name} has been played with!`);
  }
} // ✅ <-- this line is critical!

// Create your pet
const myPet = new Pet("Flocky");
console.log(myPet);

// Create an output element to show results
let output = document.createElement("div");
document.body.appendChild(output);

output.innerText = `${myPet.name} is my dog. Hunger: ${myPet.hunger}, Energy: ${myPet.energy}, Happiness: ${myPet.happiness}.`;

// Test methods
myPet.feed();
myPet.play();
```

```python
%%javascript

(function() {
  const W = 480, H = 320;
  const canvas = document.createElement('canvas');
  canvas.width = W;
  canvas.height = H;
  canvas.style.border = '1px solid #333';
  document.body.appendChild(canvas);
  const ctx = canvas.getContext('2d');

  
  class Brick {
    constructor(x,y,w,h,hits=1){
      this.x=x; this.y=y; this.w=w; this.h=h; this.hits = hits; this.alive = true;
    }
    hit(){
      this.hits -= 1;
      if(this.hits <= 0) this.alive = false;
    }
    render(ctx){
      if(!this.alive) return;
      if(this.hits===2) ctx.fillStyle = '#8B0000'; // darker (needs two hits)
      else if(this.hits===1) ctx.fillStyle = '#FFA500';
      ctx.fillRect(this.x,this.y,this.w,this.h);
      ctx.strokeStyle='#222'; ctx.strokeRect(this.x,this.y,this.w,this.h);
    }
  }

  // --- Paddle ---
  class Paddle{
    constructor(){
      this.w=75; this.h=10; this.x=(W-this.w)/2; this.y=H-30; this.speed=6;
    }
    render(ctx){
      ctx.fillStyle='#0095DD'; ctx.fillRect(this.x,this.y,this.w,this.h);
    }
    move(dir){
      this.x += dir*this.speed;
      this.x = Math.max(0, Math.min(W-this.w, this.x));
    }
  }

  // --- Ball ---
  class Ball{
    constructor(){
      this.r=8; this.x=W/2; this.y=H/2; this.angle = -Math.PI/4; this.speed = 3;
      this.updateVelocity();
    }
    updateVelocity(){
      // normalize velocity vector from angle and speed
      this.vx = Math.cos(this.angle)*this.speed;
      this.vy = Math.sin(this.angle)*this.speed;
    }
    render(ctx){
      ctx.beginPath(); ctx.arc(this.x,this.y,this.r,0,Math.PI*2); ctx.fillStyle='#00AA00'; ctx.fill(); ctx.closePath();
    }
    step(){
      this.x += this.vx; this.y += this.vy;
    }
    bounceX(){ this.vx *= -1; this.angle = Math.atan2(this.vy, this.vx); },
    bounceY(){ this.vy *= -1; this.angle = Math.atan2(this.vy, this.vx); },
    // called when ball hits the paddle - we increase speed slightly so game speeds up over time
    handlePaddleCollision(paddle){
      // reflect vertically
      this.y = paddle.y - this.r - 0.1;
      this.vy = -Math.abs(this.vy);
      // Adjust angle based on hit position
      const rel = (this.x - (paddle.x + paddle.w/2)) / (paddle.w/2);
      this.angle = rel * (Math.PI/3) - Math.PI/2;
      // *** EDITED LINE: change ball speed on paddle collision (Ball#handlePaddleCollision) ***
      this.speed *= 1.05; // increase speed by 5% on paddle hit (edited line)
      // update velocity vector to match new speed/angle
      this.updateVelocity();
    }
  }

  // --- Game ---
  class Game{
    constructor(){
      this.paddle = new Paddle();
      this.ball = new Ball();
      this.bricks = [];
      this.left=false; this.right=false;
      this.score=0; this.lives=3;
      this.populateBricks();
      this.bind();
      this.running=true;
      this.loop();
    }
    populateBricks(){
      const rows=4, cols=7; const pad=8; const bw=(W-40-(cols-1)*pad)/cols; const bh=16;
      for(let r=0;r<rows;r++){
        for(let c=0;c<cols;c++){
          const x = 20 + c*(bw+pad);
          const y = 40 + r*(bh+pad);
          // Make some bricks require two hits: e.g., alternate pattern
          const hits = ((r+c) % 3 === 0) ? 2 : 1;
          this.bricks.push(new Brick(x,y,bw,bh,hits));
        }
      }
    }
    bind(){
      window.addEventListener('keydown', e=>{ if(e.key==='ArrowLeft') this.left=true; if(e.key==='ArrowRight') this.right=true; });
      window.addEventListener('keyup', e=>{ if(e.key==='ArrowLeft') this.left=false; if(e.key==='ArrowRight') this.right=false; });
      canvas.addEventListener('click', ()=>{ if(!this.running){ this.reset(); } });
    }
    reset(){
      this.ball = new Ball(); this.paddle = new Paddle(); this.running=true; this.bricks=[]; this.populateBricks(); this.score=0; this.lives=3;
    }
    loop(){
      this.update(); this.render();
      if(this.running) requestAnimationFrame(()=>this.loop());
    }
    update(){
      // paddle movement
      if(this.left) this.paddle.move(-1);
      if(this.right) this.paddle.move(1);
      // ball step
      this.ball.step();
      // wall collisions
      if(this.ball.x - this.ball.r <= 0 || this.ball.x + this.ball.r >= W) this.ball.bounceX();
      if(this.ball.y - this.ball.r <= 0) this.ball.bounceY();
      // paddle collision
      if(this.ball.y + this.ball.r >= this.paddle.y && this.ball.y + this.ball.r <= this.paddle.y + this.paddle.h && this.ball.x >= this.paddle.x && this.ball.x <= this.paddle.x + this.paddle.w){
        this.ball.handlePaddleCollision(this.paddle);
      }
      // bottom (miss)
      if(this.ball.y - this.ball.r > H){
        this.lives -= 1;
        if(this.lives <= 0){ this.running = false; } else { this.ball = new Ball(); }
      }
      // bricks collision
      for(const b of this.bricks){
        if(!b.alive) continue;
        if(this.ball.x + this.ball.r > b.x && this.ball.x - this.ball.r < b.x + b.w && this.ball.y + this.ball.r > b.y && this.ball.y - this.ball.r < b.y + b.h){
          // simple collision response: reverse y direction
          this.ball.vy *= -1;
          // Decrease brick hits; only remove when hits reach 0
          b.hit();
          if(!b.alive) this.score += 10; else this.score += 4;
          // *** EDITED LINE: optionally slow ball slightly when hitting a brick (Ball#brickCollision) ***
          this.ball.speed = Math.max(1.5, this.ball.speed * 0.98); // slow down by 2% on brick hit (edited line)
          this.ball.updateVelocity();
          break;
        }
      }
    }
    render(){
      ctx.clearRect(0,0,W,H);
      // draw bricks
      for(const b of this.bricks) b.render(ctx);
      // paddle
      this.paddle.render(ctx);
      // ball
      this.ball.render(ctx);
      // hud
      ctx.fillStyle='#000'; ctx.font='14px sans-serif'; ctx.fillText('Score: '+this.score, 8, 16); ctx.fillText('Lives: '+this.lives, W-80,16);
      // Instructor HUD (small, bottom-left)
      ctx.fillStyle='rgba(255,255,255,0.85)'; ctx.fillRect(6, H-48, 360, 36);
      ctx.fillStyle='#000'; ctx.font='12px sans-serif';
      ctx.fillText('Instructor: Dark bricks need 2 hits. Edited speed line: Ball#handlePaddleCollision -> this.speed *= 1.05', 10, H-28);
      if(!this.running){ ctx.fillStyle='rgba(0,0,0,0.6)'; ctx.fillRect(0,0,W,H); ctx.fillStyle='#fff'; ctx.font='20px sans-serif'; ctx.fillText('Game Over - click to restart', 80, H/2); }
    }
  }

  // Start game
  new Game();
})();
```
