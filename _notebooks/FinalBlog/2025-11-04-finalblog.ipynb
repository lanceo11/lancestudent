## The Three Sprints

### Sprint 1 - Installing Tools

Installing tools in CSSE definetely was a struggle, despite being quite straight forward. I distinctly remember being close to dropping this class because my tools and VS Code were tweaking out, so I got anxious everyday because my tools wouldn't work.

This trimester taught me to be patient, learn how to problem solve, and utilize my peers and AI. Installing Jupyter Notebooks, VS Code, and setting up my GitHub account while also forking repositories became easier as the trimester went on. I entered this class wanting to learn, and I can't lie in the first month or so I was a little confused to when I would actually start to learn the basics of code.

Sprint 2 might have been the most helpful Sprint for me. In this Sprint, I laerned a lot about the basics of code such as `Classes`, `Variables`, `Booleans`, and `Strings`.

This Sprint was extremely helpful for me especially in teaching my own lesson. Despite being unable to present to the class and only to Mr. Mortensen, having to research and apply my lesson on `Strings` in example code and in a lesson format was extremely helpful in teaching me.
