## N@TM

Night at the Museum was a unique experience in the eyes of a CSSE Student. Last year, in my freshman year, I remember as a 3D Animation Student seeing some of my friends in this Computer Science classroom. I was very amazed at the amount of people present the room opposed to the other rooms. I was very interested what they were making on and I saw them presenting their games.

I was intrigued, and at the time, being curious about learning to code, N@TM was a big reason to why I ended up joining this class Sophmore year. 

When I was presenting this year, I really liked how our themes across groups really connected together. Having one big game, as opposed to different games with different themes made it feel like there was a big objective in our game.

I saw that all the parents and students coming in were interested in looking at all the other levels because each of the tinier games were connected to a bigger objective.

My group and I really liked the presence of the CSA students there at N@TM and it helped us to feel very proud of our work, especially those CSA students that took the time to play our game and add review comments.
